## Q12 — Data Completeness %

Calculate the **data completeness percentage** of the dataset.

Formula:  
`completeness = (total_values - missing_values) / total_values × 100`

Print the result rounded to 2 decimal places.

```python
# total number of cells in the DataFrame
total_values = df.size   # or df.shape[0] * df.shape[1]

total_values
```

```python
# total number of missing values in the DataFrame
missing_values = df.isna().sum().sum()

missing_values
```

```python
# completeness formula
completeness = (total_values - missing_values) / total_values * 100

# print rounded to 2 decimal places
print(f"Data completeness: {completeness:.2f}%")
```

## Q16 — Price Range Pie Chart

Plot a **pie chart** for the `Price range` column.
Label the slices as: `Cheap`, `Moderate`, `Expensive`, `Very Expensive`.
Show percentage on each slice using `autopct='%1.1f%%'`.

```python
# Map numeric price ranges to labels
price_labels = {
    1: "Cheap",
    2: "Moderate",
    3: "Expensive",
    4: "Very Expensive"
}

# Replace numbers with labels
price_series = df["Price range"].map(price_labels)

# Get counts for each category
price_counts = price_series.value_counts()

# Plot pie chart
plt.figure(figsize=(6, 6))
plt.pie(
    price_counts.values,
    labels=price_counts.index,        # Cheap, Moderate, ...
    autopct="%1.1f%%",                # show percentages
    startangle=90
)

plt.title("Price Range Distribution")
plt.show()
```

## Q25 — IQR Outlier Detection on Votes

Use the **IQR method** to detect outliers in the `Votes` column.

Steps:
1. Calculate Q1, Q3, and IQR
2. Calculate lower and upper bounds
3. Filter and print the outlier rows
4. Print the total outlier count

```python
# 1 .Calculate Q1, Q3, and IQR
Q1 = df['Votes'].quantile(0.25)
Q3 = df['Votes'].quantile(0.75)
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR", IQR)
```

```python
# 2. Calculate lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

```python
# 3. Filter and print the outlier rows
outliers = df[(df["Votes"] < lower_bound) | (df['Votes'] > upper_bound)]
print("Outlier rows in Votes column:")
outliers.head(2)
```

```python
print(len(outliers))
```

## Q31 — Challenge: City EDA Function

Write a function `city_eda(city_name)` that takes a city name and prints:
- Number of restaurants
- Average rating (exclude 0-rated)
- Most common cuisine
- % of restaurants with online delivery

Test it with `city_eda('New Delhi')`.

```python
def city_eda(city_name):
    city_df = df[df['City'] == city_name]
    print(len(city_df))

    non_zero = city_df[city_df["Aggregate rating"] > 0]
    avg_rating = non_zero["Aggregate rating"].mean()
    print(avg_rating)
    
    most_common_cuisine = city_df["Cuisines"].mode()
    if len(most_common_cuisine) > 0:
        most_common_cuisine = most_common_cuisine.iloc[0]
    else:
        most_common_cuisine = None
    
    
    if len(city_df) > 0:
        online_delivery_pct = (city_df["Has Online delivery"] == 1).mean() * 100
    else:
        online_delivery_pct = 0

    print(f"Average rating (excluding 0-rated): {avg_rating:.2f}" if not pd.isna(avg_rating) else
          "Average rating (excluding 0-rated): N/A")
    print(f"Most common cuisine: {most_common_cuisine}")
    print(f"% of restaurants with online delivery: {online_delivery_pct:.2f}%")
```

```python
city_eda('New Delhi')
```

## Q33 — Challenge: Average Rating by City (min 50 restaurants)

Filter cities that have at least 50 restaurants.
Calculate the average rating for each of those cities.
Plot a **bar chart** sorted by rating in descending order.

```python
# 1. Group by City: get count and mean rating
city_stats = df.groupby("City").agg(
    restaurant_count=("Restaurant Name", "count"),
    avg_rating=("Aggregate rating", "mean")
)
```

```python
# 2. Keep only cities with at least 50 restaurants
city_50plus = city_stats[city_stats["restaurant_count"] >= 50]
```

```python
# 3. Sort by average rating (descending)
city_50plus_sorted = city_50plus.sort_values(by="avg_rating", ascending=False)

print(city_50plus_sorted)
```

```python
# 4. Plot bar chart
plt.figure(figsize=(8, 6))
plt.bar(city_50plus_sorted.index, city_50plus_sorted["avg_rating"], color="skyblue")
plt.ylabel("Average Aggregate Rating")
plt.xlabel("City")
plt.title("Average Rating by City (Cities with ≥ 50 Restaurants)")
plt.tight_layout()
plt.show()
```
