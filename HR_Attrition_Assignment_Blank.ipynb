### Q43. Create a Heatmap of correlation between numeric columns.

```python
corr = df.corr(numeric_only=True)
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, fmt='.2f', linewidth=0.5, square=True)
plt.tight_layout()
plt.show()
```

```python
# 1. Encode Attrition as numeric (if it's Yes/No)
df_num = df.copy()
df_num["Attrition_num"] = df_num["Attrition"].map({"Yes": 1, "No": 0})

# 2. Get correlations of all numeric columns with Attrition_num
corr_all = df_num.corr(numeric_only=True)["Attrition_num"].dropna()

# 3. Pick "important" numeric columns based on a threshold
threshold = 0.1  # adjust as you like (0.15 / 0.2 for stricter)
important_cols = corr_all[abs(corr_all) >= threshold].index.tolist()

# Remove the target itself, keep only feature columns
important_cols = [c for c in important_cols if c != "Attrition_num" if c != "Attrition_binary" if c != "OverTime_binary"]

print("Important numeric columns:", important_cols)

# 4. Create correlation matrix only for these important columns
corr_imp = df_num[important_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(
    corr_imp,
    annot=True,
    fmt=".2f",
    linewidths=0.5,
    square=True
)
plt.title("Correlation Heatmap of Important Numeric Features")
plt.tight_layout()
plt.show()
```
