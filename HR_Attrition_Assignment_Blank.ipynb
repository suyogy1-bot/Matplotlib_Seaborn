### Q23. Find Top 5 Job Roles by Attrition count.

```python
top5_jobroles_attrition = (
    df[df['Attrition'] == 'Yes']
    .groupby('JobRole')['Attrition'].
    count().
    reset_index(name='AttritionCount')
)
```

```python
top5_jobroles_attrition.sort_values('AttritionCount', ascending=False).head(5)
```

## Part C - Matplotlib

### Q31. Create a Bar Chart of Department-wise employee count.

```python
Dept_counts = df['Department'].value_counts()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8, 5))
Dept_counts.plot(kind='bar')
plt.xticks(rotation=45)
plt.show()
```

### Q32. Create a Pie Chart of Attrition distribution.

```python
Att_dist = df['Attrition'].value_counts()
plt.figure(figsize=(6,4))
Att_dist.plot(kind='pie', autopct='%1.1f%%', startangle=90, counterclock=False)
plt.show()
```

### Q33. Create a Histogram of Age.

```python
age = df['Age']
plt.figure(figsize=(6,4))
plt.hist(age, bins=10, edgecolor='black')
plt.show()
```

### Q34. Create a Line Chart of average Monthly Income by Years At Company.

```python
income_by_years = df.groupby('YearsAtCompany')['MonthlyIncome'].mean().reset_index().sort_values('YearsAtCompany')
```

```python
plt.figure(figsize=(6,4))
plt.plot(income_by_years['YearsAtCompany'],
         income_by_years['MonthlyIncome'],
         marker = 'o')
plt.grid(True)
plt.show()
```

### Q35. Create a Scatter Plot of Age vs Monthly Income.

```python
plt.figure(figsize=(6,4))
plt.scatter(df['Age'], df['MonthlyIncome'], alpha=0.6)
plt.xlabel('Age')
plt.ylabel('Monthly Income')
plt.show()
```

### Q36. Create a Box Plot of Monthly Income.

```python
plt.figure(figsize=(6,4))
plt.boxplot(df['MonthlyIncome'], vert=True)
plt.show()
```

### Q37. Create a Horizontal Bar Chart of Top 10 Job Roles by average Monthly Income.

```python
jobrole_avg_income = (
    df.groupby('JobRole')['MonthlyIncome']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)
```

```python
plt.figure(figsize=(6,4))
jobrole_avg_income.plot(kind='barh')
plt.gca().invert_yaxis()  # so highest income appears at the top
plt.tight_layout()
plt.show()
```
